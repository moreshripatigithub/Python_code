## now read that volumn file

```python
df = spark.read.format("csv")\
       .option("header", "true")\
       .option("inferSchema", "true")\
       .load("/Volumes/silver/silver_schema/row_files/New Sample data.csv")

#df.display()
```

### select perticular column ---> col , selectexpr

```python
df.select(col("salary").alias("salary"))
df.select(col("salary"),col("city"),col("region"))   #.display()
```

```python
df.selectExpr(
    "emp_id",
    "salary * 12 as annual_salary"
)
```

### add a column using lit , col function

```python
df.withColumn("new_column",lit(0))   ## for integer value we can use this
df.withcolumn("new_column1",lit("yes") ## this for string column , 

df.withColumn("new_col",col("salary")**2)   ## using our column also we can do some operation
```

### rename column and hoe to drop a column

```python
df.withColumnRenamed("salary","sal").withColumnRenamed("city","city_name")
df.drop("salary").show()
df.drop("salary","city").show()
```

###sort the record


```python

# using sort

df.sort(col("salary").desc())

# using sort for multiple column

df.sort((col("salary").desc()) , col("emp_code").asc())

## using orderBy also we can  sort it
df.orderBy(col("salary").desc(),col("emp_code").asc())

```

filter the record

```python
# for single condition we can use it

df.filter(col("city")=="Mumbai")    #.display()

## for multiple condition

df.filter((col("city")=='Mumbai') & (col("region")=='W')).display()
```

### remove dublicate


```python
# remove duplicate from single column
df.dropDuplicates(["region"])  #.display()

## for multiple column remove duplicates
df.dropDuplicates(["region","city"])#.display()
```

### fill null value 


```python
#Fill all NULLs with a single value

df_filled = df.fillna("UNKNOWN")    # for stiring value
df_filled = df.fillna(0)            # for number value



## Fill NULL for specific columns

df_filled = df.fillna({
    "city": "NA",
    "salary": 0
})

```

### patturn based filter

```python
#df.where(col("city").startswith("M"))

df.where(col("city").like("M%")).display()
```

### create a column based on perticular column ---> using when condition

```python
df.withColumn("gender_sep",when(col("gender")=='Male','MALE').otherwise("Female")).display()
```

```python
df.withColumn("new_column" ,
              when(col("region")=='W','WEST').
              when(col("region")=='S','SOUTH').
              when(col("region")=='N','NORTH').
              otherwise("EAST")).display()
```
